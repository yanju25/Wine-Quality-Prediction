```python
#create a historam of the "Fixed acidity" feature
plt.figure(figsize=(9,5))
sns.histplot(data=data,x='fixed acidity',color='blue',edgecolor='black',alpha=0.5,bins=5)
plt.title("Histogram of fixed Acidity")
plt.xlabel('Fixed Acidity')
plt.ylabel('count')
plt.show()
```

```python
# calculate the MEAN of "fixed acidity" feature

round(data['fixed acidity'].mean(),2)
```

```python
# calculate the MEDIAN of "fixed acidity" feature
data['fixed acidity'].median()
```

```python
#create a historam of the "Fixed acidity" feature and also show the mean and the median
plt.figure(figsize=(9,5))
sns.histplot(data=data,x='fixed acidity',color='grey',edgecolor='black',alpha=0.5,bins=5)
plt.title("Histogram of fixed Acidity")
plt.xlabel('Fixed Acidity')
plt.ylabel('count')
plt.vlines(data['fixed acidity'].mean(),ymin=0,ymax=4000,color='purple',label='mean',linestyles='--')
plt.vlines(data['fixed acidity'].median(),ymin=0,ymax=4000,color='red',label='median')
plt.legend()
plt.show()
```

```python
#create a histogram of the "volatile acidity" feature
plt.figure(figsize=(9,6))
sns.histplot(data = data, x = 'volatile acidity', color = 'green', edgecolor ='linen',alpha = 0.9, bins = 6)
plt.title("Histogram of Volatile Acidity")
plt.xlabel('Volatil Acidity')
plt.ylabel('Count')
plt.show()
```

```python
# Plot distplot using 'Volatile acidity' feature
plt.figure(figsize=(11,6))
sns.distplot(data['volatile acidity'],color='orange')
plt.title("distplot of volatile Acidity ")
plt.xlabel('Volatile Acidity')
plt.ylabel('Density')
plt.show()
```
